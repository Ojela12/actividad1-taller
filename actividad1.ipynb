## Casos de prueba

Las funciones reciben `roles` como parámetro, así que se pueden probar configuraciones que no están en `ROLES` sin modificar `config.py`.

```python
ROLES_PRUEBA = {
    "nombre_desc": {"columnas": ["ITF", "EDAD", "ANO4"], "criterio": "nombre", "orden": "B"},
    "minimo_100": {"columnas": list(COLUMNAS), "criterio": "nombre", "orden": "A", "minimo": 100},
    "criterio_invalido": {"columnas": ["ITF", "EDAD", "CAT_OCUP"], "criterio": "promedio", "orden": "A"},
    "columna_inexistente": {"columnas": ["EDAD", "SEXO"], "criterio": "nombre", "orden": "A"},
}
for rol_prueba in ROLES_PRUEBA:
    imprimir_informe(rol_prueba, roles=ROLES_PRUEBA)
```
